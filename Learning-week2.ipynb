```python
text = "Hello, World! This is a sample text for testing purposes."
text = text.lower()
punctuation = '''!()-[]{};:'"\,<>./?@#$%^&*_~'''
for p in text:
    if p in punctuation:
        text = text.replace(p, "")
words = text.split()
word_count = {}
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1
sorted_word_count = sorted(word_count.items(), key=lambda x: x[1], reverse=True)
for word, count in sorted_word_count:
    print(f"{word}: {count}")
```

```python
student_scores = {
    "Ethan": {"math": 85, "science": 90, "english": 78},
    "Olivia": {"math": 92, "science": 88, "english": 95},
    "Liam": {"math": 78, "science": 82, "english": 80},
    "Ava": {"math": 90, "science": 85, "english": 88},
}

def add_score(student_name, subject, score):
    if student_name not in student_scores:
        student_scores[student_name] = {}
    student_scores[student_name][subject] = score
    print(f"Added score for {student_name}: {subject} = {score}")

def query_score(student_name, subject):
    if student_name in student_scores and subject in student_scores[student_name]:
        score  = student_scores[student_name][subject]
        print(f"{student_name}'s score in {subject}: {score}") 
    else:
        print(f"No score found for {student_name} in {subject}.")

def average_score(student_name):
    if student_name in student_scores:
        scores = student_scores[student_name].values()
        avg_score = sum(scores) / len(scores)
        print(f"{student_name}'s average score: {avg_score:.2f}")
    else:
        print(f"No scores found for {student_name}.")
def subject_ranking(subject):
    ranking = sorted(student_scores.items(), key=lambda x: x[1].get(subject, 0), reverse=True)
    print(f"Ranking for {subject}:")
    for rank, (student_name, scores) in enumerate(ranking, start=1):
        score = scores.get(subject, 0)
        print(f"{rank}. {student_name}: {score}")
add_score("Ethan", "history", 88)
query_score("Olivia", "math")
average_score("Liam")
subject_ranking("science")
```
