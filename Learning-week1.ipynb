```python
class Solution(object):
    def game(self, guess, answer):
        """
        :type guess: List[int]
        :type answer: List[int]
        :rtype: int
        """
        count = 0
        for g, a in zip(guess, answer):
             if g == a:
                 count += 1
        return count
```
