```python
class Solution(object):
    def game(self, guess, answer):
        """
        :type guess: List[int]
        :type answer: List[int]
        :rtype: int
        """
        count = 0
        for g, a in zip(guess, answer):
             if g == a:
                 count += 1
        return count
```

```python
def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    for i in range(3, n):
        if n % i == 0:
            return False
    return True
def main(): 
    n = int(input("Enter a number: "))
    if is_prime(n):
        print(f"{n} is a prime number.")
    else:
        print(f"{n} is not a prime number.")
```

```python
def BMI (weight, height):
    bmi = weight / (height ** 2)
    return bmi
input_weight = float(input("Enter your weight in kg: "))
input_height = float(input("Enter your height in meters: "))
bmi_value = BMI(input_weight, input_height)
if bmi_value < 18.5:
    print(f"Your BMI is {bmi_value:.2f}. You are underweight.")
elif 18.5 <= bmi_value < 24.9:
    print(f"Your BMI is {bmi_value:.2f}. You have a normal weight.")
elif 25 <= bmi_value < 29.9:
    print(f"Your BMI is {bmi_value:.2f}. You are overweight.")
else:
    print(f"Your BMI is {bmi_value:.2f}. You are obese.")  
```

```python
def leap_year (year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return True
    else:
        return False
input_year = int(input("请输入年份: "))
if leap_year(input_year):
    print(f"{input_year} 是闰年。")   
```

```python
def is_narcissistic_number(num):
    num_str = str(num)
    num_len = len(num_str)
    total = sum(int(digit) ** num_len for digit in num_str)
    return total == num
input_num = int(input("请输入一个数字: "))
if is_narcissistic_number(input_num):
    print(f"{input_num} 是水仙花数。") 
else:
    print(f"{input_num} 不是水仙花数。")
    
```

```python
def fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    
    fib_sequence = [0, 1]
    for i in range(2, n):
        next_number = fib_sequence[i - 1] + fib_sequence[i - 2]
        fib_sequence.append(next_number)
    
    return fib_sequence
input_n = int(input("请输入一个整数 n: "))
fib_sequence = fibonacci(input_n)
print(f"前 {input_n} 个斐波那契数列的数字为: {fib_sequence}")
```

```python
def dec2bin(n):
    if n == 0:
        return "0"
    result = ""
    while n > 0:
        result = str(n % 2) + result
        n //= 2
    return result
input_decimal = int(input("请输入一个十进制整数: "))
binary_representation = dec2bin(input_decimal)
print(f"{input_decimal} 的二进制表示为: {binary_representation}")
```

```python
def sum_digit(n):
    total = 0
    while n > 0:
        digit = n % 10
        total += digit
        n //= 10
    return total
input_num = int(input("请输入一个整数: "))
digit_sum = sum_digit(input_num)
print(f"{input_num} 的各位数字之和为: {digit_sum}")
```

```python
n = 4  # 控制菱形的行数，中间一行有7个星号
# 打印上半部分
i = 1
while i <= n:
    # 打印空格
    j = 1
    while j <= n - i:
        print(" ", end="")
        j += 1
    # 打印星号
    k = 1
    while k <= 2*i - 1:
        print("*", end="")
        k += 1
    print()  # 换行
    i += 1

# 打印下半部分
i = n - 1
while i >= 1:
    # 打印空格
    j = 1
    while j <= n - i:
        print(" ", end="")
        j += 1
    # 打印星号
    k = 1
    while k <= 2*i - 1:
        print("*", end="")
        k += 1
    print()
    i -= 1
```

```python
n = int(input("请输入一个正整数："))
count = 0  # 用来记录步数

while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
    count = count + 1

print(f"到达1一共走了 {count} 步")
```
